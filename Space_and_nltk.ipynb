```python
text="Dr. strange likes pav bhaji of Mumbai. Hulk likes papdi chaat of Delhi"
```

```python
import nltk
nltk.download('punkt_tab')
sentences=nltk.sent_tokenize(text)
sentences
```

```python
import spacy
nlp=spacy.load('en_core_web_sm')
doc=nlp(text)
for sentance in doc.sents:
  print(sentance)
```

```python
for sentence in doc.sents:
  for word in sentence:
    print(word)
```
