```python
import re
```

```python
chat1= 'codebasics: you ask lot of questions @ 1235678912, abc@xyz.com, 9998881234'
chat2='codebasics: here it is: (123)-567-8912, abc@xyz.com'
chat3='codebasics: yes, phone: 1235678912 email: abc@xyz.com'
```

```python
chat1
```

```python
chats = [chat1, chat2, chat3]
for chat in chats:
  print(chat)
```

```python
pattern = r"\d{10}|\(\d{3}\)-\d{3}-\d{4}/gm"
matches=re.findall(pattern, chat)
print(f"Chat: {chat}\nMatches: {matches}\n")
```

```python
pattern = r'\d{10}|\(\d{3}\)-\d{3}-\d{4}'

chats = [chat1, chat2, chat3]
for chat in chats:
  number = re.findall(pattern, chat)
  print(f"Chat: {chat}\n number: {number}\n")
```

```python
pattern=r'[a-z0-9A-Z_]*@*[a-z0-9A-Z_]*\.[a-zA-Z]*'

chats = [chat1, chat2, chat3]
for chat in chats:
  email = re.findall(pattern, chat)
  print(f"Chat: {chat}\n email: {email}\n")
```

```python
chat1='codebasics: Hello, I am having an issue with my order # 412889912'
chat2='codebasics: I have a problem with my order number 412889912'
chat3='codebasics: My order 412889912 is having an issue, I was charged 300$ when online it says 280$'
```

```python
pattern = r'order[\D]*(\d*)'

chats = [chat1, chat2, chat3]
for chat in chats:
  order_number = re.findall(pattern, chat)
  print(f"Chat: {chat}\n order_number: {order_number}\n")
```

```python
age_pattern = r'age ([\d]+)'

age=re.findall(age_pattern, text)
age
```

```python
name_pattern=r'Born(.*)'

name=re.findall(name_pattern,text)
name[0].strip()
```

```python
DOB_pattern=r'Born.*\n(.*)\('

DOB=re.findall(DOB_pattern,text)
DOB
```

```python
birthArea_pattern=r'\(age.[\d]*\)\n(.*)'

birthArea=re.findall(birthArea_pattern,text)
birthArea[0]
```

```python
def get_personal_info(text):
  age_pattern = r'age ([\d]+)'
  name_pattern=r'Born(.*)'
  DOB_pattern=r'Born.*\n(.*)\('
  birthArea_pattern=r'\(age.[\d]*\)\n(.*)'

  age=re.findall(age_pattern, text)
  name=re.findall(name_pattern,text)
  DOB=re.findall(DOB_pattern,text)
  birthArea=re.findall(birthArea_pattern,text)

  return{
      'age':int(age[0]),
      'name':name[0].strip(),
      'DOB':DOB[0],
      'BirthArea':birthArea[0]
  }
```

```python
get_personal_info(text)
```
